```python
def f(x):
    return (x + 3) ** 2
    
def gradient(x):
    return 2 * (x + 3)
    
x = 2
learning_rate = 0.1
iterations = 100

for i in range(iterations):
    x = x - learning_rate * gradient(x)
    
print("Local Minimum x =", round(x, 6))
print("Minimum Value y =", round(f(x), 6))
```
